```python
import requests
import json
```

```python
# Definir URL base da API
BASE_URL = "http://localhost:8000"
```

```python
#  (GET /canvas)
user_id = "lucas"
headers = {"user-id": user_id}

response = requests.get(f"{BASE_URL}/canvas", headers=headers)

print("Status:", response.status_code)
response.json()
```

```python
# (POST /canvas/field/{field_id})
field_id = 1
dados = {"answer": "desenvolver soluções inovadores e inteligentes para monitoramento e controle de insetos voadores."}

response = requests.post(f"{BASE_URL}/canvas/field/{field_id}", headers=headers, json=dados)

print("Status:", response.status_code)
print(response.json()['suggestion_error'])
response.json()['canvas']
```
